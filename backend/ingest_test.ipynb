```python
import pandas as pd
import os


CSV_PATH = r'D:\BigBang\backend\dataset.csv'

df = pd.read_csv(CSV_PATH, encoding='utf-8')
df["episode_name"] = df["episode_name"].str.replace('\xa0', ' ')
df.head()
```

```python
df["episode_name"][:10000].unique()
```

```python
import re

def parse_episode_info(name: str):
    pattern = r'Series (\d{2}) Episode (\d{2}) – (.+)'

    match = re.match(pattern, name)
    print(match.groups())
    if match:
        return match.groups()

    return "Unknown", "Unknown", "Unknown"
```

```python
season, episode, episode_title = parse_episode_info(
        df["episode_name"][0]
    )
print(df["episode_name"][0])
print(f"Parsed episode info: Season {season}, Episode {episode}, Title: {episode_title}")
```
